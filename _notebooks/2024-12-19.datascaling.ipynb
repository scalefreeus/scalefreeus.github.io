데이터 분석을 위해서는 원래 데이터 값을 그대로 사용하는 것보다는 적절한 형태로 변환을 해야 하는 경우가 종종 있습니다. 
데이터 변환도 데이터의 종류가 무엇이냐에 따라서 사용할 수 있는 방법이 차이가 납니다.

명목형 데이터의 경우 가장 간단한 방법은 One-Hot Encoding을 사용하는 것입니다.

```python
import pandas as pd
df = pd.read_csv('data/titanic.csv')
df.head()
```

```python
pd.get_dummies(df['Sex'])
```

기존 컬럼은 삭제하고 One-Hot Encoding된 컬럼을 추가하려면 아래와 같이 하면 됩니다.

```python
pd.get_dummies(df, columns=['Sex'])
```

또는 아래와 같이 Contingency Table을 만들 수 있습니다.

```python
pd.crosstab(index=df['Sex'], columns=df['Embarked'])
```

수치형 데이터의 경우는 정규화를 사용합니다. 만약 regression 문제를 푼다면 가능한 feature들의 값의 범위를 맞추는 것이 좋습니다. 그렇지 않으면 값이 큰 feature에 너무 크게 영향을 받게 됩니다.

```python
from sklearn.preprocessing import MinMaxScaler

# Age와 Fare의 최소값은 -1, 최대값은 1이 되도록 변환
mms = MinMaxScaler(feature_range=(-1, 1))
df[['Age_n', 'Fare_n']] = mms.fit_transform(df[['Age', 'Fare']])
```

```python
df.describe()
```

```python
from sklearn.preprocessing import StandardScaler

ss = StandardScaler()
df[['Age_s', 'Fare_s']] = ss.fit_transform(df[['Age', 'Fare']])
```

```python
df.describe()[['Age', 'Fare', 'Age_s', 'Fare_s']]
```
